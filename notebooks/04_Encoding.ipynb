In this notebook we perform Encoding
-Label Encoding
-One-Hot Encoding
-Save encoded dataset

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/processed/fraud_detection_cleaned.csv")
print("Dataset Shape :", df.shape)
```

```python
df.info()
```

```python
categorical_cols = df.select_dtypes(include="object").columns.tolist()
print("Categorical Columns")
print(categorical_cols)
```

```python
for col in categorical_cols:
    print(col)
    print(df[col].value_counts())
```

LABEL ENCODING

```python
from sklearn.preprocessing import LabelEncoder
label_encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
```

```python
df.head(3) #view encoded dataset
df.info()
```

```python
for col in categorical_cols: #verify encoding
    print(col)
    print(df[col].value_counts())
```

```python
for col in categorical_cols:    #displaying encoding mapping
    print(col)
    mapping = dict(zip(label_encoders[col].classes_,
                       label_encoders[col].transform(label_encoders[col].classes_)))
    print(mapping)
```

ONE-HOT ENCODING

```python
categorical_cols = ["payment_type","employment_status","housing_status","source","device_os"]
```

```python
df = pd.get_dummies(df,columns=categorical_cols,drop_first=True,dtype=int)
```

```python
print("Dataset Shape:", df.shape)
```

```python
encoded_cols = [col for col in df.columns if col.startswith(("payment_type_","employment_status_",
                "housing_status_","source_","device_os_"))]
print(encoded_cols)
```

```python
df.info()
```

```python
df.to_csv("../data/processed/financial_fraud_encoded.csv",index=False)
print("Encoded dataset saved successfully!")
```
