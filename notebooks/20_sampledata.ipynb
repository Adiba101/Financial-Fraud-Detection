```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/financial_fraud_feature_engineered.csv"
)

print(df.shape)
```

```python
# Create random sample
sample_df = df.sample(
    n=50000,      # Change to 30000 or 100000 if desired
    random_state=42
)

print("Sample Shape:", sample_df.shape)

# Save sample
sample_df.to_csv(
    "../data/processed/financial_fraud_feature_engineered_sample.csv",
    index=False
)

print("Sample dataset saved successfully!")
```
