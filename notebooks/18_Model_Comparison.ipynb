```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
results = {

    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "XGBoost",
        "SVM"
    ],

    "Accuracy": [
        0.7878,
        0.9438,
        0.9665,
        0.98533,
        0.93505
    ],

    "Precision": [
        0.0289,
        0.0669,
        0.1025,
        0.2021,
        0.0668
    ],

    "Recall": [
        0.5594,
        0.3164,
        0.2625,
        0.1120,
        0.3772
    ],

    "F1_Score": [
        0.0550,
        0.1104,
        0.1474,
        0.1441,
        0.1136
    ],

    "ROC_AUC": [
        0.7450,
        0.7612,
        0.8460,
        0.8540,
        0.1127      
    ]
}
comparison_df = pd.DataFrame(results)
comparison_df
```

```python
comparison_df.to_csv( "../data/processed/model_comparison.csv", index=False)
print("Model comparison saved successfully.")
```

```python
comparison_df
```

```python
colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown', 'pink', 'cyan']
plt.figure(figsize=(10,5))
plt.bar(
    comparison_df["Model"],
    comparison_df["Accuracy"],
    color=colors[:len(comparison_df)]
)
plt.xticks(rotation=45)
plt.ylabel("Accuracy")
plt.title("Model Accuracy Comparison")
plt.show()
```

```python
colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown', 'pink', 'cyan']
plt.figure(figsize=(10,5))
plt.bar(
    comparison_df["Model"],
    comparison_df["F1_Score"],
    color=colors[:len(comparison_df)]
)
plt.xticks(rotation=45)
plt.ylabel("F1_Score")
plt.title("Model  F1_Score Comparison")
plt.show()
```

```python
colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown', 'pink', 'cyan']
plt.figure(figsize=(10,5))
plt.bar(
    comparison_df["Model"],
    comparison_df["ROC_AUC"],
    color=colors[:len(comparison_df)]
)
plt.xticks(rotation=45)
plt.ylabel("ROC-AUC")
plt.title("ROC-AUC Comparison")
plt.show()
```

```python
best_model = comparison_df.loc[
    comparison_df["ROC_AUC"].idxmax()
]
print("Best Model:")
print(best_model)
```

```python
ranking = comparison_df.sort_values(
    by="ROC_AUC",
    ascending=False
)
ranking
```

```python
ranking.to_csv("../data/processed/model_ranking.csv",index=False)
print("Model ranking saved successfully.")
```
